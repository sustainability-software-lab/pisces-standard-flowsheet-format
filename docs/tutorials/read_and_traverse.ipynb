# Read and traverse an SFF file

An SFF file is a directed graph: `units` are nodes and `streams` are edges,
with `chemicals` and `utilities` as shared registries referenced by ID. This
tutorial loads `SF_BST_01.json` (corn dry-grind ethanol) as plain JSON (no validation) and
walks one edge — a stream's `source_unit_id` / `sink_unit_id` — back to the
`unit.id` values it connects.

```python
import os, json, pisces_sff
pkg = os.path.dirname(pisces_sff.__file__)
doc = json.load(open(os.path.join(pkg, "exported_flowsheets",
                     "bioindustrial_park", "SF_BST_01.json")))
print("sections:", list(doc.keys()))
print("units:", len(doc["units"]), "streams:", len(doc["streams"]),
      "chemicals:", len(doc["chemicals"]))
# Follow one edge: source_unit_id / sink_unit_id -> unit.id
s = doc["streams"][0]
print("stream", s["id"], ":", s.get("source_unit_id"), "->", s.get("sink_unit_id"))
```
