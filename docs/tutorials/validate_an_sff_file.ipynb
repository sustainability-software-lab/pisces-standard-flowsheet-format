# Validate an SFF file

SFF validation is two layers: a `jsonschema` structural gate
(`validate_json_against_schema`) and a semantic-check engine
(`validate_flowsheet_against_SFF`) implementing the requirement catalogue in
`sff_checks.md`. This tutorial runs both against the corpus's one current-shape
file, `SF_BST_01.json` (corn dry-grind ethanol).

```python
import os, collections, pisces_sff
from pisces_sff import validate_json_against_schema, validate_flowsheet_against_SFF

pkg = os.path.dirname(pisces_sff.__file__)
schema = os.path.join(pkg, "schema", "sff_schema.json")
sff = os.path.join(pkg, "exported_flowsheets", "bioindustrial_park",
                   "SF_BST_01.json")

print(validate_json_against_schema(sff, schema)[0])
ok, results = validate_flowsheet_against_SFF(sff)
print("is_valid:", ok)
print(collections.Counter((r.severity, r.status) for r in results))
```
