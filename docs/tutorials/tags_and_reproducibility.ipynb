# Tags and reproducibility

As of schema `0.1.3`, `metadata.tags` is a machine-verified array of
provenance/quality tags. There are two classes: **static** tags
(`exported-from-simulator`, `extracted-from-prose`, `extracted-from-image`),
earned by passing a tag-specific subset of the checks in `sff_checks.md` with
no `warning`-severity findings, and the **harness** tag (`reproducible`),
earned only by re-running the embedded reproducibility recipe and
deep-comparing the result.

This tutorial calls `evaluate_sff_tags` with `run_harness=False` — a static-only
check that runs no simulation.

> **Note:** earning the `reproducible` tag requires the heavy harness path
> (`verify_reproducible` / `evaluate_sff_tags(run_harness=True)`), which
> reconstructs the embedded recipe and re-runs the export inside the pinned
> conda environment. That path is shown in the Contributing and Validation
> pages and is **not** executed here.

```python
import os, pisces_sff
from pisces_sff import evaluate_sff_tags
pkg = os.path.dirname(pisces_sff.__file__)
sff = os.path.join(pkg, "exported_flowsheets", "bioindustrial_park",
                   "SF_BST_01.json")
verdict = evaluate_sff_tags(sff, run_harness=False)  # static only; no simulation
for tag, v in verdict.items():
    print(tag, "declared:", v["declared"], "earned:", v["earned"])
```
